Inventario básico

Análisis inicial de visitantes y control de inventario básico en una tienda local.

```python
# 1. Variables: producto, precio, stock
# cuaderno, Bs.10, cantidad: 15

producto_1 = "cuaderno"
precio_1 = 10
stock_1 = 15

producto_2 = "boligrafo"
precio_2 = 5
stock_2 = 20

# 2. Calcular el valor total (precio * stock)
valor_total_1 = precio_1 * stock_1
print (valor_total_1)

valor_total_2 = precio_2 * stock_2
print (valor_total_2)

total_inventario = valor_total_1 + valor_total_2

print("------------REPORTE DE INVENTARIO--------------")
print(f"Producto 1: {producto_1}, Precio: {precio_1}, Stock: {stock_1}, Valor Total: {valor_total_1}")
print(f"Producto 2: {producto_2}, Precio: {precio_2}, Stock: {stock_2}, Valor Total: {valor_total_2}")
print(f"Inventario General: {total_inventario}")
```
